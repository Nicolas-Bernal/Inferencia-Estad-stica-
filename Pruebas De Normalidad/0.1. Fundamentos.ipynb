## Graficos de curva teorica

### Para algunas distribuciones tomadas como propuestas para generar graficos

```r
set.seed(123)

X1 = rexp(1000,rate = 1)
par(mfrow = c(2,2))
hist(X1,freq = FALSE,breaks ="FD",col = "GREEN",main = "Histograma exp", xlab = "Valores", ylab = "Densidad")
curve(dexp(x,rate = 1),add = TRUE,col = "BLUE", lwd = 3 )
boxplot(X1,horizontal = TRUE,main = "Box Plot exp",col = "green")

Y1 = runif(1000,min= 0, max = 10 )
hist(Y1,freq = FALSE, breaks = "FD",col = "BLUE",main = "Histograma Unif", xlab = "Valores", ylab = "Densidad")
curve(dunif(x,min = 0,max = 10),add = TRUE,col = "GREEN",lwd = 3)
boxplot(Y1,horizontal = TRUE, main = "Box Plot Unif",col = "BLUE")
```

## Observaciones sobre los graficos

* #### En el primer histograma exp podemos notar la simetria sobre la curva teoria podemos ver que la acumulacion de los datos suceden en valores peque;os ya que tiene colas largas a la derecha los eventos suceden en peque;os momentos esto sucede en algunos casos de ocurrencia de eventos

* #### El boxplot exp se nota un poco los valores altos que aveces llega a tener con una mediana baja y igual que la mayoria de datos desplazados hacia abajo lo cual confirma una asimetria en el grafico


* #### En el histograma Unif podemos observar algo muy particular y es que a los datos estar organizados de una manera mas adecuada ya cual nos indica que en cada intervalo de tiempo sucede los mismos eventos con la misma probabilidad

* #### Ahora el boxplot lo que esta informando es una mediana perfectamente en la mitad de la caja luego una simetria sobre los bigotes de la caja ademas de confirmar la simetria perfecta del grafico

```r
X2 = rnorm(1000, mean = 5, sd = 5)

par(mfrow = c(2,2))
hist(X2,freq = FALSE, breaks = "FD",col = "yellow",main = "Histogram Binom",xlab = "Valores",ylab = "Densidad")
curve(dnorm(x,mean = 5,sd = 5),add = TRUE, lwd = 3,col = "Purple")
boxplot(X2,horizontal = TRUE,col = "yellow",main = "Box plot")

Y2 = rpois(1000, lambda = 3)
hist(Y2,freq = FALSE,breaks = "FD",col = "purple",main = "Hist Pois",xlab = "Valores",ylab = "Densidad" )

# nose podria generar una curva teoria por lo que tenemos otra alternativa

x_1 = 0:max(X2)

lines(x_1,dpois(x_1,lambda = 3),type = "h",lwd = 3,col = "yellow")
points(x_1,dpois(x_1,lambda = 3),pch = 16,col = "yellow")

boxplot(X2,horizontal = TRUE,col = "purple",main ="Box Plot")
```

## Observaciones y Conclusiones

* #### Para el histograma que se observa de unos datos normales podemos comprobar su simetria ademas de tambien poder comprobar su boxplot una simetria con una media centrada en 5 y con una desviacion estandar de 5

* #### Se observa en la distribucion de pois al ser una funcion de densidad discreta en los primeros momentos una mayor cantidad de datos acertados mientras que las barras presentan una probabilidad puntal en 1,2,3,4,... por lo que ademas el boxplot observamos por la cantidad de datos generados existen algunos atipicos como se presentan a la cola derecha
