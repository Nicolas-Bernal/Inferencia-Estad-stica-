## Pruebas De Normalidad

Existen muchas pruebas para probar que un conjunto de datos proviene de una distribucion normal o se transformo a una distribucion normal

## Shapiro-Wilk(shapiro.test)

es la prueba de normalidad mas usada para conjuntos de datos medios y peque;o

```r
datos = rnorm(200)
shapiro.test(datos)

datos2 = runif(200)
shapiro.test(datos2)

datos3 = rexp(200)
shapiro.test(datos3)
```

cuando aplicamos el test de shapiro encontramos que esta funcion evaluacia la hipotesis nula (los datos provienen de una normal) y hipotesis alternativa(los datos no provienen de una normal) el p-value es el indicador clave para tomar una decision sobre un nivel de significancia del 0.05%.

* p-value = 0.2245 > 0.05 lo que corresponde que no hay evidencia suficiente para rechazar la hispotesis nula

* W = 0.99069 es un estadistico de prueba que esta entre 0 y 1 ademas como es demasiado cercano a 1 se podria decir que es una normal

## Kolmogorov-Smirnov - lilliefors

Compara la distribucion empirica con una teorica es muy buena para conjuntos de datos muy grandes ahora cuando tenemos un conjunto de datos que no sabemos si su media o varianza y vamos a hacer estimacion de parametros usamos la version corregido de lilliefors ya que la de kolmogorov debemos ser muy exactos con la media y la desviacion estandar

```r
#install.packages("nortest")

library("nortest")

datos = rnorm(6000, mean = 50,sd = 10)

lillie.test(datos)

datos2 = rpois(6000,lambda = 10)
lillie.test(datos2)

# Ahora con Kolmogorov

ks.test(datos,"pnorm",mean = 50,sd =10)
```

analogamente el p-value es una referencia con un grado de libertad 0.05 como podemos notar algunos son mayores o menores por lo que hay es donde tomamos las decisiones ademas el p-value es diferente sobre el mismo conjunto de datos lo que cambia es la prueba que nos da mas confianzaa la de lilliefors

## Test de Anderson-Darling

Consiste en una peque;a variacion a la de Kolgomorov que lo que hace es darle mas peso a las colas de la distribucion ideal para expresar atipicos o desviaciones irregulares

```r

datos = rnorm(6000, mean = 50,sd = 10)

ad.test(datos)

datos1 = rpois(6000,lambda = 10)

ad.test(datos1)
```

Analogamente tomamos decisiones sobre el parametro p respecto si rechaza o acepta la hipotesis nula o la hipotesis alternativa
