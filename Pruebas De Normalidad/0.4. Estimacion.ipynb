## Estimacion Media y Desviacion Poblacional

#### Para este momento se empieza a dar una introduccion a las estimaciones ya que en la practica real no podemos medir una media poblacional pero podemos tomar una serie de medias muestras y con eso estimar la media poblacional en este caso se busca estimar la media para una poblacion.

#### Considere una población normal con media $\mu = 6$ y desviación estándar $\sigma = 2$. Con el fin de verificar mediante simulación las propiedades de la media muestral como estimador puntual y comprobar el Teorema del Límite Central

```r
set.seed(5238)

datos = rnorm(200*10,mean = 6,sd = 2)

muestrasnorm = matrix(datos,nrow = 200, ncol = 10)

head(muestrasnorm)

tail(muestrasnorm)

dim(muestrasnorm)
```

```r
vectormedias = apply(muestrasnorm,1,mean)
vectormedias[1:9]

par(mfrow = c(2,2))

hist(vectormedias,freq = FALSE,breaks = "FD",col = "purple", main = "Hist VectorMean")

curve(dnorm(x,mean = 6,sd = 2/sqrt(10)),add = TRUE,col = "green",lwd = 5)

boxplot(vectormedias,horizontal = TRUE,col = "purple")

acf(vectormedias,main = "ACF Vectormean")

qqnorm(vectormedias)
qqline(vectormedias)
```

```r
mean(vectormedias)

sd(vectormedias)
```

## Observaciones

* Se puede concluir que al tomar una serie de datos aleatorios y hacer pruebas sobre ellos como extrallendo medias y haciendo graficos podemos aplicar el teorema de limite central y estimar la media y desviacion poblacional

* En el histograma se observa una simetria entre la curva teorica y como se distribuyen las medias de los datos mientras en en box plot tambien se nota algo de simitria con unos datos anormales lo cual es consecuencia del tama;o de la muestra

* En el grafico acf se observa independencia salvo en un punto

* luego el qqline-norm nos esta asegurando linealidad
