### Pruebas De Normalidad

## Teorema De Limite central

## Enunciado

#### sea, $X_1,...,X_n$ ,muestras aleatorias independientes y identicamente distribuidas con media y varianza.
<br>
$$E(X_n) = μ$$
<br>

<br>
$$Var(X_i) = σ^2<∞$$
<br>

#### Entonces, cuando $n → ∞$ la variable
<br>
$$ \frac{\bar{x} - μ }{ σ / \sqrt{n} } → N(0,1)$$
<br>


### Observaciones

analizando la parte superior del teorema donde $\bar{X}-μ$ es la diferencia que mide cuanto se aleja el promedio observado del real o poblacional.

Ahora con el denominador analogamente se conoce el error de la desviacion estandar analogamente o error estandar  

## Aplicacion

Usando el Teorema del Límite Central, generar múltiples muestras aleatorias de tamaño n provenientes de una distribución exponencial, calcular sus medias muestrales y verificar empíricamente que la distribución de dichas medias se aproxima a una distribución normal.

```r
mediasexp_20 = c()
for (k in 1:1000){
    mediasexp_20[k] = mean(rexp(20,rate = 2))

}
hist(mediasexp_20,freq = FALSE, breaks = "FD", col = "purple", main = "1. Histograma De Medias ")
curve(dnorm(x,mean = 1/2,sd = sqrt((1/(2^2))/20)),add = TRUE,col = "green",lwd = 3)
```

```r
# 1. Parámetros de la simulación
n1 = 20
n2 = 100
lambda = 2

set.seed(123)
mediasexp_20  <- replicate(1000, mean(rexp(n1, rate = lambda)))
mediasexp_100 <- replicate(1000, mean(rexp(n2, rate = lambda)))

par(mfrow = c(2,1))


hist(mediasexp_20, freq = FALSE, breaks = "FD", col = "purple",main = "1. Histograma De Medias (n = 20)")
curve(dnorm(x, mean = 1/lambda, sd = sqrt((1/lambda^2)/n1)), add = TRUE, col = "green", lwd = 3)

hist(mediasexp_100, freq = FALSE, breaks = "FD", col = "purple", main = "2. Histograma De Medias (n = 100)")
curve(dnorm(x, mean = 1/lambda, sd = sqrt((1/lambda^2)/n2)), add = TRUE, col = "green", lwd = 3)
```

### Observaciones

* Se puede observar una gran similitud entre la curva teoria de una normal con media y varianza dada esto a la naturaleza de los residuos

* como se observa en la continuacion de los graficos podemos concluir que al aumentar el tama;o de la muestra conseguimos un mejoramiento en la simetria de esos residuos
