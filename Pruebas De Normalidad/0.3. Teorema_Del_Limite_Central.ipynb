### Pruebas De Normalidad

## Teorema De Limite central

## Enunciado

#### sea, $X_1,...,X_n$ ,muestras aleatorias independientes y identicamente distribuidas con media y varianza.
<br>
$$E(X_n) = μ$$
<br>

<br>
$$Var(X_i) = σ^2<∞$$
<br>

#### Entonces, cuando $n → ∞$ la variable
<br>
$$ \frac{\bar{x} - μ }{ σ / \sqrt{n} } → N(0,1)$$
<br>


### Observaciones

analizando la parte superior del teorema donde $\bar{X}-μ$ es la diferencia que mide cuanto se aleja el promedio observado del real o poblacional.

Ahora con el denominador analogamente se conoce el error de la desviacion estandar analogamente o error estandar  

## Aplicacion

Usando el Teorema del Límite Central, generar múltiples muestras aleatorias de tamaño n provenientes de una distribución exponencial, calcular sus medias muestrales y verificar empíricamente que la distribución de dichas medias se aproxima a una distribución normal.

```r
mediasexp_20 = c()
for (k in 1:1000){
    mediasexp_20[k] = mean(rexp(20,rate = 2))

}
hist(mediasexp_20,freq = FALSE, breaks = "FD", col = "purple", main = "1. Histograma De Medias ")
curve(dnorm(x,mean = 1/2,sd = sqrt((1/(2^2))/20)),add = TRUE,col = "green",lwd = 3)
```

```r
# 1. Parámetros de la simulación
n1 = 20
n2 = 100
lambda = 2

set.seed(123)
mediasexp_20  <- replicate(1000, mean(rexp(n1, rate = lambda)))
mediasexp_100 <- replicate(1000, mean(rexp(n2, rate = lambda)))

par(mfrow = c(2,1))


hist(mediasexp_20, freq = FALSE, breaks = "FD", col = "purple",main = "1. Histograma De Medias (n = 20)")
curve(dnorm(x, mean = 1/lambda, sd = sqrt((1/lambda^2)/n1)), add = TRUE, col = "green", lwd = 3)

hist(mediasexp_100, freq = FALSE, breaks = "FD", col = "purple", main = "2. Histograma De Medias (n = 100)")
curve(dnorm(x, mean = 1/lambda, sd = sqrt((1/lambda^2)/n2)), add = TRUE, col = "green", lwd = 3)
```

### Observaciones

* Se puede observar una gran similitud entre la curva teoria de una normal con media y varianza dada esto a la naturaleza de los residuos

* como se observa en la continuacion de los graficos podemos concluir que al aumentar el tama;o de la muestra conseguimos un mejoramiento en la simetria de esos residuos

## Aplicacion

Genere 60 muestras de tamaño 40 de una población Poisson de media 3 (con semilla igual a 4321). Calcule el valor del estadístico media muestral para cada muestra. A continuación conteste las siguientes cuestiones:

* ¿Cuál es el valor de la media muestral de la primera y quinta muestra que ha generado?

* Represente la muestra de 60 valores del estadístico X mediante un histograma.

* ¿Qué distribución asintótica sigue el estadístico X?

* ¿Cuál es la media muestral de las observaciones de X?

* Calcule la cuasidesviación típica muestral y la desviación típica muestral de las observaciones de X

* Comente conjuntamente las tres últimas preguntas.

* Dibuje la distribución asintótica del estadístico X sobre el histograma de la muestra de 60 valores del estadístico. Comente el gráfico

```r
set.seed(4321)

datos = rpois(60*40, lambda = 3)
muestrapois = matrix(datos,nrow = 60,ncol = 40)

dim(muestrapois)

# Vector De Medias

vectormedia = apply(muestrapois,1,mean)
vectormedia

# Valor de la primera y quinta muestra

vectormedia[1]
vectormedia[5]

# Hist

hist(vectormedia,breaks = "FD",freq = FALSE,main = "Histograma Medias Muestrales", col = "pink")
```

Aun que los Datos originales provienen de una distribucion discreta podemos ver que las medias muestrales se comportan aproximadamente como una normal cuando el tama;o de la muestra es suficientemente grande en algunos casos

para esto podemos usar el teorema de limite central

```r
par(mfrow = c(2,2))

hist(vectormedia,breaks = "FD",freq = FALSE,main = "Histograma Medias Muestrales", col = "pink")
curve(dnorm(x, mean = 3, sd = sqrt(3/40)),add = TRUE, col = "red", lwd = 3)
qqnorm(vectormedia)
qqline(vectormedia)
acf(vectormedia,main = "ACF")
boxplot(vectormedia,horizontal = TRUE,col = "pink",main = "Box Plot")
```

De lo anterior podemos ver la gran similitud entre la curva teorica y las medias muestrales luego con el qqplot se observa una simetria entre la linea mientras tampoco se ve datos irregulares en el acf ni box plot  y una simetria se puede confirmar normalidad
